# Implementing a Stack Class

First, we define a class `Stack` with its methods defined in an interface. We also define a *constructor* for this class.  The constructor takes an uninitialized, empty object `S` 
and initializes its *member variable* `mStackElements`. This member variable is a list containing the data stored in the stack.

```typescript
class Stack {
  mStackElements: any[];
  constructor() {
    this.mStackElements = [];
  }
}
interface Stack {
  mStackElements: any[];
  push(item: any): void;
  pop(): any;
  top(): any;
  isEmpty(): boolean; 
  copy(): Stack;
  toString(): string;
  convert(): string;
}
```

```typescript
const S = new Stack();
console.log(S);
```

Next, we add the method `push` to the class `Stack`.  The method $\texttt{push}(S, e)$ pushes $e$ onto the stack $S$. We add this method to the class `Stack` through modifying the prototype of the class Stack.

```typescript
Stack.prototype.push = function(this: MyStack, e: any): void {
    this.mStackElements.push(e);
};
```

The method `pop` removes the topmost element from a stack.  It is an error to `pop` an empty stack.

```typescript
Stack.prototype.pop = function(this: Stack) {
    if (this.mStackElements.length === 0) {
        throw new Error("popping empty stack");
    }
    this.mStackElements.pop();
};
```

The method `top` returns the element that is on top of the stack.
It is an error to call this method if the stack is empty.

```typescript
Stack.prototype.top = function(this: MyStack) {
    if (this.mStackElements.length === 0) {
        throw new Error("top of empty stack");
    }
    return this.mStackElements[this.mStackElements.length - 1];
};
```

The method `S.isEmpty()` checks whether the stack `S` is empty.

```typescript
Stack.prototype.isEmpty = function(this: Stack) {
    return this.mStackElements.length === 0;
};
```

The method `S.copy()` creates a *shallow* copy of the given stack, i.e. the copy contains the same objects as
the stack `S`.

```typescript
Stack.prototype.copy = function(this: Stack) {
    const C = new Stack();
    C.mStackElements = [...this.mStackElements];
    return C;
};
```

The method `S.toString()` converts a stack `S` into a string. 

```typescript
console.log(['ab', 'cd'].join('\n'));
```

```typescript
Stack.prototype.toString = function(this: Stack) {
    const C = this.copy();
    const result = C.convert();
    const dashes = "-".repeat(result.length);
    return [dashes, result, dashes].join("\n");
};
```

The method `convert` converts a stack into a string.

```typescript
Stack.prototype.convert = function(this: Stack): string {
    if (this.isEmpty()) { 
        return '|';
    }
    const top = this.top();
    this.pop();            
    return this.convert() + ' ' + String(top) + ' |';
};
```

The method `createStack(L)` takes a list `L` and pushes all of its elements on a newly created stack, which is
then returned.

```typescript
function createStack(L: any[]): Stack {
    const S = new Stack();
    for (const x of L) {
        S.push(x);
        console.log(S.toString());
    }
    return S;
}
```
