```typescript
class Complex {
  public mComment?: string;
  constructor(public mRe: number, public mIm: number) {}

  add(other: Complex): Complex {
    return new Complex(this.mRe + other.mRe, this.mIm + other.mIm);
  }

  sub(other: Complex): Complex {
    return new Complex(this.mRe - other.mRe, this.mIm - other.mIm);
  }

  mul(other: Complex): Complex {
    return new Complex(
      this.mRe * other.mRe - this.mIm * other.mIm,
      this.mRe * other.mIm + this.mIm * other.mRe
    );
  }

  div(other: Complex): Complex {
    const d = other.mRe ** 2 + other.mIm ** 2;
    return new Complex(
      (this.mRe * other.mRe + this.mIm * other.mIm) / d,
      (this.mIm * other.mRe - this.mRe * other.mIm) / d
    );
  }

  toString(): string {
    return `${this.mRe} + i * ${this.mIm}`;
  }
}
```

```typescript
const z1 = new Complex(0, 1);
z1.toString();
```

```typescript
const z2 = new Complex(3, 4);
z1.mul(z2).toString();
```

```typescript
const z = new Complex(1, 5);
z.toString();
```
