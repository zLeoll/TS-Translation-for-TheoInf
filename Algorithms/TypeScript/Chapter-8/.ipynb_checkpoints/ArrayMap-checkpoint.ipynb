# Implementing Maps as Arrays

If the keys are natural numbers less than a given natural number `n` that is not 
too large, a map can be implemented using an array.
The class `ArrayMap` shows how this is done.  The constructor `constructor(n: number)` takes one argument `n`, which specifies the size of the internal array.

```typescript
class ArrayMap<T> {
  private mArray: (T | null)[];

  constructor(n: number) {
    this.mArray = Array(n).fill(null);
  }

  find(k: number): T | null {
    return this.mArray[k];
  }

  insert(k: number, v: T): void {
    this.mArray[k] = v;
  }

  delete(k: number): void {
    this.mArray[k] = null;
  }

  toString(): string {
    let result = "{ ";
    this.mArray.forEach((value, key) => {
      if (value !== null) {
        result += `${key}: ${value}, `;
      }
    });
    if (result === "{ ") {
      return "{}";
    }
    return result.slice(0, -2) + " }";
  }
}
```

```typescript
const squares = new ArrayMap<number>(10);
console.log(squares.toString());
```

```typescript
for (let i = 0; i < 10; i++) {
  squares.insert(i, i * i);
}
console.log(squares.toString());
```

```typescript
for (let i = 0; i < 10; i++) {
  console.log(`find(${i}) = ${squares.find(i)}`);
}
```

```typescript
for (let i = 0; i < 10; i++) {
  squares.delete(i);
}
console.log(squares.toString());
```

Lets compute the primes up to $n = 10000$.

```typescript
const n = 10000;
```

```typescript
console.time("Compute primes");

const S = new ArrayMap<number>(n + 1);
for (let i = 2; i <= n; i++) {
  S.insert(i, i);
}

for (let i = 2; i <= n / 2 + 1; i++) {
  for (let j = i; j <= n; j++) {
    if (i * j > n) break;
    S.delete(i * j);
  }
}

console.timeEnd("Compute primes");
```

```typescript
let primes: number[] = [];
for (let k = 2; k <= n; k++) {
  if (S.find(k) !== null) {
    primes.push(k);
  }
}
console.log(primes.join(" "));
```
