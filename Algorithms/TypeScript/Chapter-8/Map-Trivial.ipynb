# A Trivial Implementation of the ADT <em style="color:blue;">Map</em>

The following trivial implementation of the ADT <em style="color:blue;">Map</em> is based on *JavaScript’s* [options](https://developer.mozilla.org/en-US/docs/Web/JavaScript/Reference/Global_Objects/Object).

```typescript
class MapADT<K, V> {
  private mDictionary: Record<string, V>;

  constructor() {
    this.mDictionary = {};
  }

  find(k: string): V | undefined {
    return this.mDictionary[k];
  }

  insert(k: string, v: V): void {
    this.mDictionary[k] = v;
  }

  delete(k: string): void {
    delete this.mDictionary[k];
  }

  toString(): string {
    return JSON.stringify(this.mDictionary);
  }
}
```

```typescript
const m = new MapADT<string, number>();
m.insert("a", 1);
m.insert("b", 2);
m.insert("c", 3);
m.toString();
```
