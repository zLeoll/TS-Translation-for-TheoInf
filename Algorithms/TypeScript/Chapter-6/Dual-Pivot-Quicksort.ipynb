# Dual Pivot Quicksort

In this section we implement **Dual Pivot Quicksort** in *TypeScript*.

The algorithm is divided into two parts:

1. The helper function `partition`, which divides an array into three subarrays based on two pivot elements.
2. The recursive function `sort`, which uses `partition` to perform the sorting.

The function partition receives three arguments:
* `p1` and `p2` are comparable objects taken from some array that is to be sorted.
  Furthermore `p1 <= p2`.
* `L` is a list of comparable elements. 
The function *partions* the array into three subarrays.  Mathematically, we have:
$$ \texttt{partition}(p_1, p_2, L) = \bigl\langle[x \in L \mid x < p_1],\; [x \in L \mid p_1 \leq x \leq p_2],\; [x \in L \mid p_2 < x]\bigr\rangle $$

```typescript
function partition(p1: number, p2: number, L: number[]): [number[], number[], number[]] {
  if (L.length === 0) {
    return [[], [], []];
  }

  const [x, ...R] = L;
  const [R1, R2, R3] = partition(p1, p2, R);

  if (x < p1) {
    return [[x, ...R1], R2, R3];
  } else if (x <= p2) {
    return [R1, [x, ...R2], R3];
  } else {
    return [R1, R2, [x, ...R3]];
  }
}
```

The function `sort` takes an array `L` that is to be sorted and returns the sorted array.
It implements *dual pivot quicksort* and uses the first and the last element as *pivots* `p1` and `p2`.
This algorithm is specified as follows:
1. $\texttt{len}(L) \leq 1 \rightarrow \texttt{sort}(L) = L$.
2. If  $\texttt{len}(L) > 1$ we define:
   * $x := L[0]$, $y := L[-1]$,
   * $p_1 := \min(x, y)$, $p_2 := \max(x, y)$,
   * $S := [z \in L[1\!:\!-1] \mid z < p_1]$,
   * $M := [z \in L[1\!:\!-1] \mid p_1 \leq z \leq p_2]$, and
   * $B := [z \in L[1\!:\!-1] \mid p_2 < z]$.
   
   Then there are two cases:
   + $p_1 < p_2 \rightarrow \texttt{sort}(L) = \texttt{sort}(S) + [p_1] + \texttt{sort}(M) + [p_2] + \texttt{sort}(B)$,
   + $p_1 = p_2 \rightarrow \texttt{sort}(L) = \texttt{sort}(S) + [p_1] + M + [p_2] + \texttt{sort}(B)$.
      
     For the efficiency of the algorithm it is important that we do not have to sort the array $M$ in the second case. 

```typescript
function sort(L: number[]): number[] {
  if (L.length <= 1) {
    return L;
  }

  const x = L[0];
  const y = L[L.length - 1];
  const R = L.slice(1, -1);

  const p1 = Math.min(x, y);
  const p2 = Math.max(x, y);

  const [L1, L2, L3] = partition(p1, p2, R);

  if (p1 === p2) {
    return [...sort(L1), p1, ...L2, p2, ...sort(L3)];
  } else {
    return [...sort(L1), p1, ...sort(L2), p2, ...sort(L3)];
  }
}
```

```typescript
partition(5, 13, [1, 19, 27, 2, 5, 6, 4, 7, 8, 5, 8, 17, 13]);
```

```typescript
sort([1, 19, 27, 2, 5, 6, 4, 7, 8, 5, 8, 17, 13]);
```
