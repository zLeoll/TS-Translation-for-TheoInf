# Merge Sort: A Recursive, Array Based Implementation

```typescript
let L = [1, 2, 3];
let A = [...L]; 
A[2] = 7;
```

```typescript
console.log("L =", L); 
console.log("A =", A); 
```

The function `merge` takes five arguments.
  - `L`      is an array,
  - `start`  is an integer such that $\texttt{start}  \in \{0, \cdots, \texttt{len}(L)-1 \}$,
  - `middle` is an integer such that $\texttt{middle} \in \{0, \cdots, \texttt{len}(L)-1 \}$,
  - `end`    is an integer such that $\texttt{end}    \in \{0, \cdots, \texttt{len}(L)-1 \}$, 
  - `A`      is an array of the same length as `L`.
  
Furthermore, the indices `start`, `middle`, and `end` have to satisfy the following:
$$ 0 \leq \texttt{start} < \texttt{middle} < \texttt{end} \leq \texttt{len}(L) $$
The function assumes that the subarrays `L[start:middle]` and `L[middle:end]` are already sorted.
The function merges these subarrays so that when the call returns the subarray `L[start:end]`
is sorted.  The last argument `A` is used as auxiliary memory.

```typescript
function merge(L: number[], start: number, middle: number, end: number, A: number[]): void {
  for (let k = start; k < end; k++) {
    A[k] = L[k];
  }

  let idx1 = start;
  let idx2 = middle;
  let i = start;

  while (idx1 < middle && idx2 < end) {
    if (A[idx1] <= A[idx2]) {
      L[i] = A[idx1];
      idx1++;
    } else {
      L[i] = A[idx2];
      idx2++;
    }
    i++;
  }

  if (idx1 < middle) {
    for (let k = idx1; k < middle; k++, i++) {
      L[i] = A[k];
    }
  }

  if (idx2 < end) {
    for (let k = idx2; k < end; k++, i++) {
      L[i] = A[k];
    }
  }
}
```

The function `mergeSort` is called with 4 arguments.  
- The first parameter $\texttt{L}$ is the array that is to be sorted. However, the task of $\texttt{mergeSort}$ is not to sort the entire array $\texttt{L}$ but only
  the part of $\texttt{L}$ that is given as  
  $$
  \texttt{L[start:end]}
  $$
- Hence, the parameters $\texttt{start}$ and $\texttt{end}$ are indices specifying the 
  subarray that needs to be sorted.  
- The final parameter $\texttt{A}$ is used as an auxiliary array. This array is needed
  as <em style="color:blue">temporary storage</em> and is required to have the same size as the array $\texttt{L}$.

```typescript
function mergeSort(L: number[], start: number, end: number, A: number[]): void {
  if (end - start < 2) {
    return; 
  }
  const middle = Math.floor((start + end) / 2);
  mergeSort(L, start, middle, A);
  mergeSort(L, middle, end, A);
  merge(L, start, middle, end, A);
}
```

The function $\texttt{sort}(L)$ sorts an array $L$ in place using <em style="color:blue">merge sort</em>.
The function `sort` is a wrapper for the function `mergeSort`.  Its sole purpose is to allocate the auxiliary array `A`, 
which has the same size as the array holding `L`.

```typescript
function sort(L: number[]): void {
  const A = [...L];
  mergeSort(L, 0, L.length, A);
}
```

## Testing

The function `counter` takes an array as input and returns a Map that keeps count of how many times each item occurs in the array.

```typescript
function counter<T>(arr: T[]): Map<T, number> {
  const counts = new Map<T, number>();
  for (const item of arr) {
    counts.set(item, (counts.get(item) ?? 0) + 1);
  }
  return counts;
}
```

```typescript
console.log(counter(['a', 'b', 'a', 'b', 'c', 'a']));
```

We also define the helper function `compareCounter` to be able to compare the contents of two counters.

```typescript
function compareCounters(a: Map<number, number>, b: Map<number, number>): boolean {
  if (a.size !== b.size) return false;
  for (const [key, value] of a) {
    if (b.get(key) !== value) return false;
  }
  return true;
}
```

```typescript
function demo() {
  const L: number[] = Array.from({ length: 19 }, () => Math.floor(Math.random() * 99) + 1);
  console.log("L =", L);

  let S = [...L];
  sort(S);
  console.log("S =", S);

  const counterL = counter(L);
  const counterS = counter(S);

  console.log(counterL);
  console.log(counterS);
  const equal = compareCounters(counterL, counterS);
  console.log(equal);
}
```

```typescript
demo();
```

The function `isOrdered(L)` checks that the array `L` is sorted in ascending order.

```typescript
function isOrdered(L: number[]): void {
  for (let i = 0; i < L.length - 1; i++) {
    if (L[i] >= L[i + 1]) {
      throw new Error(`${L} not ordered at ${i}`);
    }
  }
}
```

The function `sameElements(L, S)` returns `true` if the arrays `L` and `S` contain the same elements and, furthermore, each 
element $x$ occurring in `L` occurs in `S` the same number of times it occurs in `L`.

```typescript
import assert from 'assert';

function sameElements(L: number[], S: number[]): void {
  assert(compareCounters(counter(L), counter(S)), "L and S do not have the same elements");
}
```

The function $\texttt{testSort}(n, k)$ generates $n$ random arrays of length $k$, sorts them, and checks whether the output is sorted and contains the same elements as the input.

```typescript
function testSort(n: number, k: number): void {
  for (let i = 0; i < n; i++) {
    const L: number[] = Array.from({ length: k }, () => Math.floor(Math.random() * (2 * k)));
    const oldL = [...L];
    sort(L);
    isOrdered(L);
    sameElements(oldL, L);
    process.stdout.write('.');
  }
  console.log();
  console.log("All tests successful!");
}
```

```typescript
console.time();
testSort(100, 20000);
console.timeEnd();
```
