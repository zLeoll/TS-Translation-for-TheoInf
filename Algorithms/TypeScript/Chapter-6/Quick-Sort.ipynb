# Quicksort

The basic idea of <em style="color:blue;">quick sort</em> is as follows:
  - If the array $\mathtt{L}$ that is to be sorted is empty, we return $\mathtt{L}$: 
    $$ \mathtt{sort}([]) = [] $$
  - Otherwise, we have $\mathtt{L} = [\mathtt{x}] + \mathtt{R}$.  In this case, we split $\mathtt{R}$ into two arrays 
    $\mathtt{S}$ and $\mathtt{B}$.
     - The array $\mathtt{S}$ (the letter $\mathtt{S}$ stands for $\underline{s}\textrm{mall}$) contains all those elements 
       of $\mathtt{R}$ that are less than or equal to $\mathtt{x}$, we have:
       $$ \mathtt{S} := [\mathtt{y} \in \mathtt{R} \mid \mathtt{y} \leq \mathtt{x}] $$
     - The array $\mathtt{B}$ (the letter $\mathtt{B}$ stands for $\underline{b}\textrm{ig}$) contains those elements of $\mathtt{R}$ 
       that are bigger than $\mathtt{x}$, we have:
       $$ \mathtt{B} := [\mathtt{y} \in \mathtt{R} \mid \mathtt{y} > \mathtt{x}] $$
   
    The process of splitting the array $\mathtt{R}$ into the arrays $\mathtt{S}$ and $\mathtt{B}$
    is called <em style="color:blue;">partitioning</em>.  After partitioning the array $\mathtt{R}$ into the arrays
    $\mathtt{S}$ and $\mathtt{B}$, these arrays are sorted
    recursively.  Then, the result is computed by putting $\mathtt{x}$ between the arrays $\mathtt{sort}(\mathtt{S})$ 
    and $\mathtt{sort}(\mathtt{B})$:
    $$\mathtt{sort}([\mathtt{x}]+\mathtt{R}) = \mathtt{sort}(\mathtt{S}) + [\mathtt{x}] + \mathtt{sort}(\mathtt{B})$$

```typescript
function sort(L: number[]): number[] {
  if (L.length === 0) return [];
  const [x, ...R] = L;
  const S = R.filter(y => y <= x);
  const B = R.filter(y => y >  x);
  return [...sort(S), x, ...sort(B)];
}
```

```typescript
sort([7, 8, 11, 12, 2, 5, 3, 7, 9, 3, 2])
```
