# Selection Sort

We define the helper function `removeFirst(x, L)` recursively. There are three cases:

- If `L` is empty, the call would be an error, because we assume that `x` occurs in `L`:
  
  $$
  removeFirst(x, []) \text{ is an error}
  $$

- If `x` equals the first element of `L`, i.e., `L = [x] + R`, we return the rest:

  $$
  removeFirst(x, [x] + R) = R
  $$

- Otherwise, keep the head `y` and remove `x` recursively from the rest:

  $$
  x \ne y \;\Rightarrow\; removeFirst(x, [y] + R) = [y] + removeFirst(x, R)
  $$

> Note: We use the name `removeFirst(x, L)` instead of `delete(x, L)` because `delete` is a reserved keyword in TypeScript.

```typescript
function minOf(L: number[]): number {
  if (L.length === 0) {
    throw new Error("minOf([]) is undefined");
  }
  return Math.min(...L);
}

function removeFirst(x: number, L: number[]): number[] {
  if (L.length === 0) {
    throw new Error(`removeFirst(${x}, [])`);
  }
  const [y, ...R] = L;
  if (y === x) {
    return R;
  }
  return [y, ...removeFirst(x, R)];
}
```

The algorithm is specified by two equations for an array `L`:

- If `L` is empty, the result is the empty array:
    $$ \mathtt{sort}([]) = [] $$
- Otherwise, we compute the smallest element of `L`, remove its **first** occurrence,
  sort the remainder recursively, and place the smallest element in front:
    $$ L \not= [] \rightarrow \mathtt{sort}\bigl(L\bigr) = 
       \bigl[\texttt{min}(L)\bigr] + \mathtt{sort}\bigl(\mathtt{removeFirst}(\texttt{min}(L), L)\bigr)
    $$

```typescript
function sort(L: number[]): number[] {
  if (L.length === 0) {
    return [];
  }
  const m = minOf(L);
  return [m, ...sort(removeFirst(m, L))];
}
```

```typescript
const L = [3, 5, 7, 4, 8, 1, 2, 3, 11, 13, 2];
sort(L);
```
