### Combining Expressions and Input Processing

Next, we show how it is possible to combine multiple expressions in a script and how to handle input data.

To begin with, we present a script that defines a natural number $n$ as a string (simulating user input) and then proceeds to compute the sum of all natural numbers from $1$ up to and including $n$. That is, it computes the sum:

$$\sum_{i=1}^{n} i = 1 + 2 + \dots + n$$

The script contained in the cell below works as follows:

* The variable `inputString` holds the input value as a string.
* This string is then converted into an integer using the function `parseInt`.
* Next, the array `L` is created such that it contains the sequence of numbers from $1$ to $n$.
* The `console.log` statement uses the `reduce` method to compute the sum of all elements of the list `L`.

````typescript
const inputString = "10";
const n = parseInt(inputString, 10);
const L = Array.from({ length: n }, (_, i) => i + 1);
const total = L.reduce((acc, curr) => acc + curr, 0);
console.log(`The sum 1 + 2 + ... + ${n} is equal to ${total}.`);
````

If we input the number $36$, which happens to be equal to $6^2$, then we discover the following remarkable identity:
$$ \sum\limits_{i=1}^{6^2} i = 666. $$

The next example shows how functions can be defined in TypeScript. The function `sum` is supposed to compute the sum of all the numbers in the set $\{0, \dots, n\}$. Therefore, we have

$$\text{sum}(n) = \sum_{i=0}^n i$$

The function `sum` is defined recursively. The recursive implementation of the function `sum` can best be understood if we observe that it satisfies the following two equations:

$$\text{sum}(0) = 0$$
$$\text{sum}(n) = \text{sum}(n-1) + n, \quad \text{provided that } n > 0.$$

The keyword `function` is used to signal that we define a function.

````typescript
function sum(n: number): number {
    if (n === 0) {
        return 0;
    }
    return sum(n - 1) + n;
}
````

Lets test this function.

````typescript
sum(10);
````

Let us discuss the implementation of the function `sum` line by line:

* The keyword `function` starts the definition of the function.
* It is followed by the name of the function that is defined. The name is followed by the list of the parameters of the function. This list is enclosed in parentheses. Unlike Python, we specify the types of the parameters and the return type.
* The body of the function is enclosed in curly braces `{ }`.
* The first statement of the body is a conditional statement, which starts with the keyword `if`. The condition `n === 0` is enclosed in parentheses.
* The next line contains a return statement.
* The last line of the function definition contains the recursive invocation of the function `sum` that computes the value `sum(n)` by reducing it to the value `sum(n-1)`.

Using the function `sum`, we can compute the sum for a given input as follows:

````typescript
const inputVal = "10";
const num = parseInt(inputVal, 10);
const total = sum(num);

if (num > 2) {
    console.log(`0 + 1 + 2 + ... + ${num} = ${total}`);
} else {
    console.log(total);
}
````

## Boolean Values and Boolean Operators

In *TypeScript*, the *truth values*, also known as *Boolean values*, are written as `true` and `false` (lowercase).

The following function is needed for pretty-printing. It assumes that the argument `val` is a truth value. This truth value is then turned into a string that has a size of exactly 5 characters.

````typescript
function toStr(val: boolean): string {
    if (val) {
        return 'True ';
    }
    return 'False';
}
````

These values can be combined using the *Boolean operators* $\wedge$, $\vee$, and $\neg$. In *TypeScript*, these operators are denoted as `&&` (AND), `||` (OR), and `!` (NOT). The following table shows how the operator `&&` is defined:

````typescript
let B = [true, false];
for (let x of B) {
    for (let y of B) {
        console.log(toStr(x), 'and', toStr(y), '=', x && y);
    }
}
````

The *disjunction* of two Boolean values is only *false* if both values are *false*. In *TypeScript*, the disjunction operator is written as `||`:

````typescript
for (let x of B) {
    for (let y of B) {
        console.log(toStr(x), 'or', toStr(y), '=', x || y);
    }
}
````

Finally, the negation operator `!` works as expected:

````typescript
for (let x of B) {
    console.log('not', toStr(x), '=', !x);
}
````

* $a\;\texttt{>=}\;b$ is true iff $a$ is bigger than or equal to $b$.
* $a\;\texttt{>}\;b$ is true iff $a$ is bigger than $b$.

## Control Structures

First of all, *TypeScript* supports *branching* via the keywords `if`, `else if`, and `else`. The following example demonstrates this logic:

````typescript
let x = 17; 

if (x < 0) {
    console.log('The number is negative!');
} else if (x === 0) {
    console.log('The number is zero.');
} else if (x === 1) {
    console.log("It's a one.");
} else {
    console.log("It's more than one.");
}
````

*Loops* can be created with the keyword `for`. They can be used to iterate over arrays. 
The syntax of a `for...of`-loop is:

```typescript
for (let x of L) {
    S1
    ...
    Sn
}
```

Here, `x` is the name of a variable that is used to iterate over the elements of the array `L`. The statements
`S1`, $\cdots$, `Sn` make up the *body* of the loop. These statements are executed for every element of the array `L`.

The next examples shows a simple loop that prints all the elements from the array `L`:

````typescript
for (let x of L) {
    console.log(x);
}
````

Another way to write a *loop* is to use the keyword `while` as in the following example:

````typescript
x = 1;
while (x < 10) {
    console.log(x);
    x = x + 1;
}
````

The general syntax of a `while`-loop is as follows:

```typescript
while (e) {
    S1
    ...
    Sn
}
```

Here, `while` is the keyword starting the `while`-loop, while `e` is an expression that contains some variable `x` which is presumably updated in the statements `S1`, $\cdots$, `Sn`. These statements are executed as long as the expression `e` is `true`.

The following program computes the prime numbers according to an
[algorithm given by Eratosthenes](https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes).
 1. We set $n$ equal to 100 as we want to compute the set all prime numbers less or equal that 100.
 2. `primes` is the array of numbers from 0 upto $n$, i.e. we have initially
 $ \texttt{primes} = [0,1,2,\cdots,n] $
 Therefore, we have initially
 $$ \texttt{primes}[i] = i \quad \mbox{for all $i \in \{0,1,\cdots,n\}$.} $$
 The idea is to set $\texttt{primes}[i]$ to zero iff $i$ is a proper product of two numbers.
 3. To this end we iterate over all $i$ and $j$ from the array $[2,\cdots,n]$
 and set the product $\texttt{primes}[i\cdot j]$ to zero. 
 This is achieved by the two `for` loops in line 4 and line 5 below.

 Note that we have to check that the product $i * j$ is not bigger than $n$ 
 for otherwise we would get an *out of range error* (or simply extend the array unexpectedly in JS) when trying to assign $\texttt{primes}[i*j]$. 
 
 4. After the iteration, all non-prime elements greater than one of the array primes have been set to zero.
 5. Finally, we compute the array of all prime numbers by collecting those elements from the array `primes` 
 that have not been set to $0$.

````typescript
let n = 1_000_000;
let primes = Array.from({length: n + 1}, (_, i) => i);
primes[1] = 0;

for (let i = 2; i <= n; i++) {
    for (let j = 2; j <= n; j++) {
        if (i * j <= n) {
            primes[i * j] = 0;
        }
    }
}
let result = primes.filter(p => p !== 0 && p >= 2);
console.log(result.length);
````

The algorithm given above can be improved by using the following observations:
 1. If a number $x$ can be written as a product $a \cdot b$, then at least one of the numbers 
 $a$ or $b$ has to be less than $\sqrt{x}$. Therefore, the `for` loop in line 5 below 
 iterates as long as $i \leq \sqrt{x}$. The function `Math.ceil` is needed to cast the square 
 root of $x$ to a natural number. 
 2. When we iterate over $j$ in the inner loop, it is sufficient if we start with $j = i$ 
 since all products of the form $i \cdot j$ where $j < i$ have already been eliminated at the time 
 when the multiples of $i$ had been eliminated.
 3. If $\texttt{primes}[i] = 0$, then $i$ is not a prime and hence it has to be a product of two 
 numbers $a$ and $b$ both of which are smaller than $i$. However, since all the multiples of 
 $a$ and $b$ have already been eliminated, there is no point in eliminating the multiples of 
 $i$ since these are also multiples of both $a$ and $b$ and hence they have already been eliminated. 
 Therefore, if $\texttt{primes}[i] = 0$ we can immediately jump to the next value of $i$. 
 This is achieved by the *continue* statement in line 8 below. 

The program shown below is easily capable of computing all prime numbers less than a million.
On my desktop computer, which is a 2017 iMac with a Quad-Core Intel i5 processor, this takes less than a second.

````typescript
Math.sqrt(10)
````

````typescript
Math.ceil(Math.sqrt(10))
````

````typescript
let n = 1_000_000;
let primes = Array.from({length: n + 1}, (_, i) => i);
let limit = Math.ceil(Math.sqrt(n));

console.time("primes")
for (let i = 2; i <= limit; i++) {
    if (primes[i] === 0) {
        continue;
    }
    let j = i;
    while (i * j <= n) {
        primes[i * j] = 0;
        j++;
    }
}
let Result = primes.filter(p => p !== 0 && p >= 2);

console.timeEnd("primes")
console.log(Result.length);
````
