# Sets implemented as AVL Trees

This notebook implements <em style="color:blue;">sets</em> as <a href="https://en.wikipedia.org/wiki/AVL_tree">AVL trees</a>.  The set $\mathcal{A}$ of <em style="color:blue;">AVL trees</em> is defined inductively:

  - $\texttt{Nil} \in \mathcal{A}$.
  - $\texttt{Node}(k,l,r) \in \mathcal{A}\quad$  iff 
      - $\texttt{Node}(k,l,r) \in \mathcal{B}_<$,
      - $l, r \in \mathcal{A}$, and
      - $|l.\texttt{height}() - r.\texttt{height}()| \leq 1$.

According to this definition, an AVL tree is an <em style="color:blue;">ordered binary tree</em>
such that for every node $\texttt{Node}(k,l,r)$ in this tree the height of the left subtree $l$ and the right
subtree  $r$ differ at most by one. 

The class `Set` represents the nodes of an AVL tree.  This class has the following member variables:

  - `mKey` is the key stored at the root of the tree,
  - `mLeft` is the left subtree, 
  - `mRight` is the right subtree, and
  - `mHeight` is the height.

The constructor `constructor()` creates the empty tree.

```typescript
class Set<T>{
    static sNodeCount = 0;
    mKey: T | null;
    mLeft: Set<T> | null;
    mRight: Set<T> | null;
    mHeight: number;
    mID?: number;
    constructor() {
        this.mKey = null;
        this.mLeft = null;
        this.mRight = null;
        this.mHeight = 0;
    }
}
interface Set<T> {
    mKey: T | null;
    mLeft: Set<T> | null;
    mRight: Set<T> | null;
    mHeight: number;
    isEmpty(): boolean;
    insert(key: T): Set<T>;
    delete(key: T): Set<T>;
    pop(): T;
    member(key: T): boolean;
    delMin(): Set<T>;
    _restore(): void;
    _update(t: Set<T>): void;
    _delMin(): [Set<T>, T | null];
    _setValues(k: T | null, l: Set<T> | null, r: Set<T> | null): void;
    _restoreHeight(): void;
    _assignIDs(NodeDict: Record<number, Set<T>>): void;
    toList(): T[];
    toString(): string;
    toDot(): Promise<void>;
}
```

Given an ordered binary tree $t$, the expression $t.\texttt{isEmpty}()$ checks whether $t$ is the empty tree.

```typescript
Set.prototype.isEmpty = function(): boolean {
    return this.mKey === null;
};
```

Given an ordered binary tree $t$ and a key $k$, the expression $t.\texttt{member}(k)$ returns `true` if the key $k$ is stored in the tree $t$.
The method `member` is defined inductively as follows:
  - $\texttt{Nil}.\texttt{member}(k) = \Omega$,

    because the empty tree is interpreted as the empty map.
  - $\texttt{Node}(k, l, r).\texttt{member}(k) = v$,
      
    because the node $\texttt{Node}(k,l,r)$ stores the assignment $k \mapsto v$.
  - $k_1 < k_2 \rightarrow \texttt{Node}(k_2, l, r).\texttt{member}(k_1) = l.\texttt{member}(k_1)$,

    because if $k_1$ is less than $k_2$, then any mapping for $k_1$ has to be stored in the left subtree  $l$.
  - $k_1 > k_2 \rightarrow \texttt{Node}(k_2, l, r).\texttt{member}(k_1) = r.\texttt{member}(k_1)$,

    because if $k_1$ is greater than $k_2$, then any mapping for $k_1$ has to be stored in the right subtree  $r$.

```typescript
Set.prototype.member = function<T>(this: Set<T>, key: T): boolean {
    if (this.isEmpty()) {
        return false;
    } else if (this.mKey === key) {
        return true;
    } else if (key < this.mKey) {
        return this.mLeft.member(key);
    } else {
        return this.mRight.member(key);
    }
};
```

The method  $\texttt{insert}()$ is specified via recursive equations.
  - $\texttt{Nil}.\texttt{insert}(k) = \texttt{Node}(k, \texttt{Nil}, \texttt{Nil})$,
  - $\texttt{Node}(k, l, r).\texttt{insert}(k) = \texttt{Node}(k, l, r)$,
  - $k_1 < k_2 \rightarrow 
          \texttt{Node}(k_2, l, r).\texttt{insert}(k_1) =
          \texttt{Node}\bigl(k_2, l.\texttt{insert}(k_1), r\bigr).\texttt{restore}()$,
  - $k_1 > k_2 \rightarrow 
         \texttt{Node}(k_2, l, r).\texttt{insert}\bigl(k_1\bigr) = 
         \texttt{Node}\bigl(k_2, l, r.\texttt{insert}(k_1)\bigr).\texttt{restore}()$.

The function $\texttt{restore}$ is an auxiliary function that is defined below.  This function restores the balancing condition if it is violated after an insertion.

```typescript
Set.prototype.insert = function<T>(this: Set<T>, key: T): Set<T> {
    if (this.isEmpty()) {
        this.mKey = key;
        this.mLeft = new Set<T>();
        this.mRight = new Set<T>();
        this.mHeight = 1;
    } else if (this.mKey === key) {
        // do nothing, key already in set
    } else if (key < this.mKey!) {
        this.mLeft.insert(key);
        this._restore();
    } else {
        this.mRight.insert(key);
        this._restore();
    }
    return this;
};
```

The method $\texttt{delete}(k)$ removes the key $k$ from the tree.  It is defined as follows:

  - $\texttt{Nil}.\texttt{delete}(k) = \texttt{Nil}$,
  - $\texttt{Node}(k,\texttt{Nil},r).\texttt{delete}(k) = r$,
  - $\texttt{Node}(k,l,\texttt{Nil}).\texttt{delete}(k) = l$,
  - $l \not= \texttt{Nil} \,\wedge\, r \not= \texttt{Nil} \,\wedge\, 
       \langle r',k_{min} \rangle := r.\texttt{delMin}()  \;\rightarrow\;
      \texttt{Node}(k,l,r).\texttt{delete}(k) = \texttt{Node}(k_{min},l,r')$
  - $k_1 < k_2 \rightarrow \texttt{Node}(k_2,l,r).\texttt{delete}(k_1) = 
    \texttt{Node}\bigl(k_2,l.\texttt{delete}(k_1),r\bigr)$,
  - $k_1 > k_2 \rightarrow \texttt{Node}(k_2,l,r).\texttt{delete}(k_1) = 
     \texttt{Node}\bigl(k_2,l,r.\texttt{delete}(k_1)\bigr)$.

```typescript
Set.prototype.delete = function<T>(this: Set<T>, key: T): Set<T> {
    if (this.isEmpty()) {
        return this;
    }
    if (key === this.mKey) {
        if (this.mLeft.isEmpty()) {
            this._update(this.mRight);
        } else if (this.mRight.isEmpty()) {
            this._update(this.mLeft);
        } else {
            const [newRight, minKey] = this.mRight._delMin();
            this.mRight = newRight;
            this.mKey = minKey;
        }
    } else if (key < this.mKey) {
        this.mLeft.delete(key);
    } else {
        this.mRight.delete(key);
    }
    return this;
};
```

The method $\texttt{delMin}()$ removes the smallest key from the given tree
and returns a pair of the form
$$ (\texttt{this}, k_m) $$
where $\texttt{this}$ is the tree that remains after removing the smallest key, while $k_m$ is the smallest key that has been found.  

The function is defined as follows:

  - $\texttt{Node}(k, \texttt{Nil}, r).\texttt{delMin}() = \langle r, k \rangle$,
  - $l\not= \texttt{Nil} \wedge \langle l',k_{min}\rangle := l.\texttt{delMin}() 
       \;\rightarrow\;
     \texttt{Node}(k, l, r).\texttt{delMin}() = 
     \langle \texttt{Node}(k, l', r).\texttt{restore}(), k_{min} \rangle
    $

```typescript
Set.prototype._delMin = function<T>(this: Set<T>): [Set<T>, T | null] {
    if (this.mLeft!.isEmpty()) {
        return [this.mRight!, this.mKey]; // Gibt rechten Teilbaum + Schlüssel zurück
    } else {
        const [ls, km] = this.mLeft!._delMin();
        this.mLeft = ls;
        this._restore();
        return [this, km];
    }
};
```

Given two ordered binary trees $s$ and $t$, the expression $s.\texttt{update}(t)$ overwrites the attributes of $s$ with the corresponding attributes of $t$.

```typescript
Set.prototype._update = function<T>(this: Set<T>, t: Set<T>): void {
    this.mKey = t.mKey;
    this.mLeft = t.mLeft;
    this.mRight = t.mRight;
    this.mHeight = t.mHeight;
};
```

The function $\texttt{createNode}(k, l, r)$ creates an AVL-tree of that has the key $k$ stored at its root, 
left subtree $l$ and right subtree $r$.

```typescript
function createNode<T>(key: T, left: Set<T>, right: Set<T>): Set<T> {
    const node = new Set<T>();
    node.mKey = key;
    node.mLeft = left;
    node.mRight = right;
    node.mHeight = Math.max(left.mHeight, right.mHeight) + 1;
    return node;
}
```

The function $\texttt{restore}(\texttt{t}his)$ restores the balancing condition of the given binary tree
at the root node and recompute the variable $\texttt{mHeight}$.

The method $\texttt{restore}$ is specified via conditional equations.

  - $\texttt{Nil}.\texttt{restore}() = \texttt{Nil}$,

    because the empty tree already is an  AVL tree.
  - $|l.\texttt{height}() - r.\texttt{height}()| \leq 1 \rightarrow 
       \texttt{Node}(k,l,r).\texttt{restore}() = \texttt{Node}(k,l,r)$.

      If the balancing condition is satisfied, then nothing needs to be done. 
  - $\begin{array}[t]{cl}
              & l_1.\texttt{height}() = r_1.\texttt{height}() + 2    \\ 
       \wedge & l_1 = \texttt{Node}(k_2,l_2,r_2)                 \\
       \wedge & l_2.\texttt{height}() \geq r_2.\texttt{height}()     \\[0.2cm]
       \rightarrow & \texttt{Node}(k_1,l_1,r_1).\texttt{restore}() = 
                     \texttt{Node}\bigl(k_2,l_2,\texttt{Node}(k_1,r_2,r_1)\bigr)
       \end{array}
    $
  - $\begin{array}[t]{cl}
               & l_1.\texttt{height}() = r_1.\texttt{height}() + 2    \\ 
        \wedge & l_1 = \texttt{Node}(k_2,l_2,r_2)               \\
        \wedge & l_2.\texttt{height}() < r_2.\texttt{height}()     \\
        \wedge & r_2 = \texttt{Node}(k_3,l_3,r_3)               \\
        \rightarrow & \texttt{Node}(k_1,l_1,r_1).\texttt{restore}() = 
                      \texttt{Node}\bigl(k_3,\texttt{Node}(k_2,l_2,l_3),\texttt{Node}(k_1,r_3,r_1) \bigr)
        \end{array}
    $
  - $\begin{array}[t]{cl}
              & r_1.\texttt{height}() = l_1.\texttt{height}() + 2    \\ 
       \wedge & r_1 = \texttt{Node}(k_2,l_2,r_2)               \\
       \wedge & r_2.\texttt{height}() \geq l_2.\texttt{height}()     \\[0.2cm]
       \rightarrow & \texttt{Node}(k_1,l_1,r_1).\texttt{restore}() = 
                     \texttt{Node}\bigl(k_2,\texttt{Node}(k_1,l_1,l_2),r_2\bigr)
       \end{array}
    $
  - $\begin{array}[t]{cl}
               & r_1.\texttt{height}() = l_1.\texttt{height}() + 2    \\ 
        \wedge & r_1 = \texttt{Node}(k_2,l_2,r_2)               \\
        \wedge & r_2.\texttt{height}() < l_2.\texttt{height}()     \\
        \wedge & l_2 = \texttt{Node}(k_3,l_3,r_3)               \\
        \rightarrow & \texttt{Node}(k_1,l_1,r_1).\texttt{restore}() = 
                      \texttt{Node}\bigl(k_3,\texttt{Node}(k_1,l_1,l_3),\texttt{Node}(k_2,r_3,r_2) \bigr)
        \end{array}
    $

```typescript
Set.prototype._restore = function<T>(this: Set<T>): void {
    if (Math.abs(this.mLeft!.mHeight - this.mRight!.mHeight) <= 1) {
        this._restoreHeight();
        return;
    }
    if (this.mLeft.mHeight > this.mRight.mHeight) {
        const k1 = this.mKey;
        const l1 = this.mLeft;
        const r1 = this.mRight;
        const k2 = l1.mKey;
        const l2 = l1.mLeft;
        const r2 = l1.mRight;
        if (l2.mHeight >= r2.mHeight) {
            this._setValues(k2, l2, createNode(k1!, r2, r1));
        } else {
            const k3 = r2.mKey;
            const l3 = r2.mLeft;
            const r3 = r2.mRight;
            this._setValues(k3, createNode(k2, l2, l3), createNode(k1, r3, r1));
        }
    } else if (this.mRight!.mHeight > this.mLeft!.mHeight) {
        const k1 = this.mKey;
        const l1 = this.mLeft;
        const r1 = this.mRight;
        const k2 = r1.mKey;
        const l2 = r1.mLeft;
        const r2 = r1.mRight;
        if (r2.mHeight >= l2.mHeight) {
            this._setValues(k2, createNode(k1, l1, l2), r2);
        } else {
            const k3 = l2.mKey;
            const l3 = l2.mLeft;
            const r3 = l2.mRight;
            this._setValues(k3, createNode(k1, l1, l3), createNode(k2, r3, r2));
        }
    }
    this._restoreHeight();
};
```

The function $\texttt{_setValues}(k, l, r)$ overwrites the member variables of the node $\texttt{self}$ with the given values.

```typescript
Set.prototype._setValues = function<T>(this: Set<T>, k: T, l: Set<T>, r: Set<T>): void {
    this.mKey = k;
    this.mLeft = l;
    this.mRight = r;
};
```

```typescript
Set.prototype._restoreHeight = function<T>(this: Set<T>): void {
    this.mHeight = Math.max(
        this.mLeft ? this.mLeft.mHeight : 0,
        this.mRight ? this.mRight.mHeight : 0
    ) + 1;
};
```

The method $t.\texttt{pop}()$ take an AVL tree $t$ and removes and returns the smallest key that is present in $t$.  It is specified as follows:
  - $\texttt{Nil}.\texttt{pop}() = \Omega$
  - $\texttt{Node}(k,\texttt{Nil}, r).\texttt{pop}() = \langle k, r\rangle$
  - $l \not=\texttt{Nil} \wedge \langle k',l'\rangle := l.\texttt{pop}() \rightarrow
     \texttt{Node}(k, l, r).\texttt{pop}() = \langle k', \texttt{Node}(k, l', r)\rangle$

```typescript
Set.prototype.pop = function<T>(this: Set<T>): T {
    if (this.mKey === null) {
        throw new Error('KeyError: pop from empty set');
    }
    if (this.mLeft!.mKey === null) {
        const key = this.mKey;
        this._update(this.mRight!);
        return key;
    }
    return this.mLeft!.pop();
};
```

## Display Code

```typescript
import { Graphviz } from "@hpcc-js/wasm";
import { display } from "tslab";
```

Given an ordered binary tree, this function renders the tree graphically using `graphviz`.

```typescript
Set.prototype.toDot = async function<T>(this: Set<T>): Promise<void> {
    Set.sNodeCount = 0;
    const NodeDict: Record<number, Set<T>> = {};
    this._assignIDs(NodeDict);
    let dot = 'digraph {\n';
    for (const [n, t] of Object.entries(NodeDict)) {
        if (t.mKey !== null) {
            dot += `  ${n} [label="${t.mKey}"]\n`;
        } else {
            dot += `  ${n} [label="", shape="point"]\n`;
        }
    }
    for (const [n, t] of Object.entries(NodeDict)) {
        if (t.mLeft && t.mLeft.mID !== undefined) {
            dot += `  ${n} -> ${t.mLeft.mID}\n`;
        }
        if (t.mRight && t.mRight.mID !== undefined) {
            dot += `  ${n} -> ${t.mRight.mID}\n`;
        }
    }
    dot += '}';
    const gv = await Graphviz.load();
    const svg = gv.layout(dot, "svg", "dot");
    const widthMatch = svg.match(/width="([^"]+)"/);
    const heightMatch = svg.match(/height="([^"]+)"/);
    const formattedWidth = widthMatch[0].replace(/"/g, '').replace('=', ':');
    const formattedHeight = heightMatch[0].replace(/"/g, '').replace('=', ':');
    const html = `
            <div style='${formattedWidth}; ${formattedHeight};'>
                ${svg}
            </div>
    `;
    display.html(html);
};
```

This method assigns a unique identifier with each node. The record `NodeDict` maps these identifiers to the nodes where they occur.

```typescript
Set.prototype._assignIDs = function<T>(this: Set<T>, NodeDict: Record<number, Set<T>>): void {
    Set.sNodeCount += 1;
    this.mID = Set.sNodeCount;
    NodeDict[this.mID] = this;
    if (this.isEmpty()) {
        return;
    }
    this.mLeft._assignIDs(NodeDict);
    this.mRight._assignIDs(NodeDict);
};
```

## Testing

The function $\texttt{demo}()$ creates a small ordered binary tree.

```typescript
function demo(): Set<string> {
    const m = new Set<string>();
    m.insert("anton");
    m.insert("hugo");
    m.insert("gustav");
    m.insert("jens");
    m.insert("hubert");
    m.insert("andre");
    m.insert("philipp");
    m.insert("rene");
    return m;
}
```

```typescript
const t = demo();
await t.toDot();
```

```typescript
while (!t.isEmpty()) {
    console.log(t.pop());
    await t.toDot();
}
```

Let's generate an ordered binary tree with random keys. 

The function `randomIntRange(min, max)` generates a random integer within the range `[min, max)`. The lower bound `min` is included, while the upper bound `max` is excluded.

```typescript
function randomIntRange(min: number, max: number): number {
  return Math.floor(Math.random() * (max - min)) + min;
}
```

```typescript
const t = new Set<number>();
for (let k = 0; k < 30; k++) {
    const randomValue = randomIntRange(0, 100);
    t.insert(randomValue);
}
await t.toDot();
let output = '';
while (!t.isEmpty()) {
    output += t.pop() + ' ';
}
console.log(output);
await t.toDot();
```

This tree looks more or less balanced.  Lets us try to create a tree by inserting sorted numbers because that resulted in linear complexity for ordered binary trees.

```typescript
const t = new Set<number>();
for (let k = 0; k < 30; k++) {
    t.insert(k);
}
await t.toDot();
let output = '';
while (!t.isEmpty()) {
    output += t.pop() + ' ';
}
console.log(output);
await t.toDot();
```

Next, we compute the set of prime numbers $\leq 100$.  Mathematically, this set is given as follows:
$$ \bigl\{2, \cdots, 100 \bigr\} - \bigl\{ i \cdot j \bigm| i, j \in \{2, \cdots, 100 \}\bigr\}$$

```typescript
const S = new Set<number>();
for (let k = 2; k <= 100; k++) {
    S.insert(k);
}
await S.toDot();
for (let i = 2; i <= 100; i++) {
    for (let j = 2; j <= 100; j++) {
        S.delete(i * j);
    }
}
await S.toDot();
let output = '';
while (!S.isEmpty()) {
    output += S.pop() + ' ';
}
console.log(output);
await S.toDot();
```
