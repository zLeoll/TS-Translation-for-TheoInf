# A Naive Implementation of the Union-Find Algorithm

Given a partition $\mathcal{P}$ of a set $M$ and an element $x$ from $M$, the function $\texttt{find}(x, \mathcal{P})$ 
returns the subset $S \in \mathcal{P}$ such that $x \in S$.

```typescript
function find(x: number, P: Set<number>[]): Set<number> | undefined {
  return P.find(S => S.has(x));
}
```

Given a set $M$ and a binary relation $R \subseteq M \times M$, the function $\texttt{union\_find}$ returns a partition $\mathcal{P}$ of $M$ such that we have
$$ \forall \langle x, y \rangle \in R: \exists S \in \mathcal{P}: \bigl(x \in S \wedge y \in S\bigr) $$
It works by starting with the trivial partition 
$$\mathcal{P} := \bigl\{ \{ x \} \mid x \in M \bigr\}. $$
As long as there is a pair $\langle x, y \rangle \in R$ such that:
* $x \in A$, $y \in B$, where $A \in \mathcal{P}$ and $B \in \mathcal{P}$, 
* but $A \not= B$

we update the partition $\mathcal{P}$ as follows:
$$
\mathcal{P} := \mathcal{P} \setminus \{ A, B \} \cup \{ A \cup B \}.
$$

Since frozen sets are not supported, the partition $\mathcal{P}$  is implemented as an array, which makes this implementation quite inefficient.

```typescript
function union_find(
  M: number[],
  R: [number, number][]
): Set<number>[] {
  console.log("R =", R);

  const P: Set<number>[] = M.map(x => new Set([x]));
  console.log("P =", P.map(s => Array.from(s)));

  function find(x: number, P: Set<number>[]): Set<number> | undefined {
    return P.find(S => S.has(x));
  }

  for (const [x, y] of R) {
    const A = find(x, P);
    const B = find(y, P);
    if (A && B && A !== B) {
      console.log(`${x} ≡ ${y}: combining ${Array.from(A)} and ${Array.from(B)}`);
      P.splice(P.indexOf(A), 1);
      P.splice(P.indexOf(B), 1);
      P.push(new Set([...A, ...B]));
      console.log("P =", P.map(S => Array.from(S)));
    }
  }
  return P;
}
```

```typescript
function demo() {
  const M = Array.from({ length: 9 }, (_, i) => i + 1);
  const R: [number, number][] = [
    [1, 4],
    [7, 9],
    [3, 5],
    [2, 6],
    [5, 8],
    [1, 9],
    [4, 7],
  ];
  const P = union_find(M, R);
}
```
