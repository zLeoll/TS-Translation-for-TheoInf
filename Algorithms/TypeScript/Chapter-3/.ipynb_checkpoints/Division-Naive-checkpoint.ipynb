# Euclidean Division

The function $\texttt{divMod}(a, n)$ takes two natural numbers $a, n \in \mathbb{N}$ such that $n > 0$ and 
returns a pair $(q, r)$ where $q$ is the <em style="color:blue">quotient</em> of dividing $a$ by $n$, 
while $r$ is the <em style="color:blue">remainder</em>.
Mathematically, $q$ and $r$ are defined as those number that satisfy the following:
  - $a = q \cdot n + r$
  - $0 \leq r < n$

```typescript
function divMod(a: number, n: number): [number, number] {
    let q = 0;
    while(a >= n){
        a -= n;
        q += 1;
    }
    return [q, a];
}
```

```typescript
divMod(12,3)
```

```typescript
divMod(14, 5)
```

```typescript
function testDivMod(m: number, n: number): void {
  const [q, r] = divMod(m, n);
  console.assert(
    m === q * n + r,
    `m = ${m}, n = ${n}, q = ${q}, r = ${r}`
  );
  console.assert(
    0 <= r && r < n,
    `m = ${m}, n = ${n}, q = ${q}, r = ${r}`
  );
}
```

```typescript
function getRandomInt(max: number): number {
  return Math.floor(Math.random() * max);
}
```

```typescript
console.time("divModLoopTime");
for (let k = 0; k < 10_000; k++) {
  const m = getRandomInt(1_000_000);
  const n = getRandomInt(1_000) + 1;
  testDivMod(m, n);
}
console.timeEnd("divModLoopTime");
```
