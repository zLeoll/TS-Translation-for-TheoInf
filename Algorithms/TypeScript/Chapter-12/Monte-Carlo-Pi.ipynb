# Computing [𝜋](https://en.wikipedia.org/wiki/Pi) with the Monte-Carlo-Method

```typescript
function randomInRange(minIncl: number, maxIncl: number): number {
  return minIncl + (maxIncl - minIncl) * Math.random();
}
```

The unit circle $U$ is defined as the set 
$$U := \bigl\{ (x,y) \in \mathbb{R}^2 \bigm| x^2 + y^2 \leq 1 \bigr\}.$$
The set $U$ contains those points $(x,y)$ that have distance of $1$ or less from the origin
$(0,0)$.  The unit circle is a subset of the square $Q$ that is defined as 
$$Q := \bigl\{ (x,y) \in \mathbb{R}^2 \bigm| -1 \leq x \leq +1 \wedge -1 \leq y \leq +1 \bigr\}.$$
A simple algorithm to compute $\pi$ works as follows:  We randomly choose $n$ points $(x,y) \in Q$.
Then we count the number of points that end up in the unit circle $U$.  Call this number $k$.
It is reasonable to assume that approximately $k$ is to $n$ as the area of the unit circle $U$ 
is to the area of the square $Q$.  As the area of $Q$ is
$2 \cdot 2$ and the area of $U$ equals $\pi \cdot 1^2$, we should have
$$\frac{k}{n} \approx \frac{\pi}{4}.$$
Multiplying by $4$ we get
$$\pi \approx 4 \cdot \frac{k}{n}.$$
The function $\texttt{approximate\_pi}(n)$ creates $n$ random points in $Q$ and approximates $\pi$ as $\displaystyle 4 \cdot \frac{k}{n}$.

```typescript
function approximatePi(n: number): number {
  let k = 0;
  for (let i = 0; i < n; i++) {
    const x = randomInRange(-1, 1);
    const y = randomInRange(-1, 1);
    const r2 = x * x + y * y;
    if (r2 <= 1) k += 1;
  }
  return 4 * (k / n);
}
```

The method $\texttt{confidence\_interval}(k, n)$ runs $k$ approximations of $\pi$ using $n$ trials in each approximation run.
It computes a $97.3\%$ confidence interval for the value of $\pi$.

```typescript
console.time("Monte-Carlo π (100..10,000,000)");
let n = 100;

while (n <= 10_000_000) {
  const pi = approximatePi(n);
  const err = Math.abs(pi - Math.PI);
  console.log(`${String(n).padStart(9)}: π = ${pi.toFixed(6)}, error: ${err.toFixed(6)}`);
  n *= 10;
}

console.timeEnd("Monte-Carlo π (100..10,000,000)");
```
