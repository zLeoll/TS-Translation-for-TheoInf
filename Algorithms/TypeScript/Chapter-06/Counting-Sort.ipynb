# Counting Sort

The procedure `countingSort` receives one arguments:
- `Students` is an array of pairs of the form `[name, grade]` where
  * `name`  is the name of a student and
  * `grade` is the grade of this student.
  
  The array `Student` is sorted alphabetically with respect to the student names.

The procedure `countingSort` returns a new array of pairs where the students
are sorted with respect to their grade. Within the same grade, students remain sorted alphabetically by name.

```typescript
type Student = [string, number];

function countingSort(Students: Student[]): Student[] {
  const maxGrade = 255;
  const Counts = new Array<number>(maxGrade + 1).fill(0);
  const Index = new Array<number>(maxGrade + 1).fill(0);
  const Sorted = new Array<Student | null>(Students.length).fill(null);

  for (const [_, grade] of Students) {
    Counts[grade]++;
  }

  Index[0] = 0;
  for (let grade = 0; grade < maxGrade; grade++) {
    Index[grade + 1] = Index[grade] + Counts[grade];
  }

  for (const [name, grade] of Students) {
    const idx = Index[grade];
    Sorted[idx] = [name, grade];
    Index[grade]++;
  }

  return Sorted as Student[];
}
```

```typescript
const Students: [string, number][] = [
  ["Alexander", 4],
  ["Benjamin", 2],
  ["Daniel", 3],
  ["David", 3],
  ["Elijah", 2],
  ["Gabriel", 1],
  ["Henry", 2],
  ["Jacob", 5],
  ["James", 3],
  ["Joseph", 2],
  ["Liam", 2],
  ["Logan", 3],
  ["Lucas", 1],
  ["Mason", 2],
  ["Matthew", 5],
  ["Michael", 3],
  ["Noah", 4],
  ["Oliver", 2],
  ["Owen", 4],
  ["Samuel", 3],
  ["Sebastian", 2],
  ["William", 1],
];
```

```typescript
function main(): void {
  const Sorted = countingSort(Students);

  for (const [name, grade] of Sorted) {
    console.log(`${name.padEnd(9)} : ${grade}`);
  }
}
```

```typescript
main();
```
