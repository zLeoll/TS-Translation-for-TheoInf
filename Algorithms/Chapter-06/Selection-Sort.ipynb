# Selection Sort

We define the helper function `remove(x, L)` recursively. There are three cases:

- If `L` is empty, the call would be an error, because we assume that `x` occurs in `L`:
  
  $$
  \mathtt{remove}(x, []) \text{ is an error}
  $$

- If `x` equals the first element of `L`, i.e., `L = [x] + R`, we return the rest:

  $$
  \mathtt{remove}(x, [x] + R) = R
  $$

- Otherwise, keep the head `y` and remove `x` recursively from the rest:

  $$
  x \ne y \;\Rightarrow\; \mathtt{remove}(x, [y] + R) = [y] + \mathtt{remove}(x, R)
  $$


```typescript
function remove(x: number, L: number[]): number[] {
  console.assert(L.length !== 0, `remove(${x}, [])`);
  const [y, ...R] = L;
  if (y === x) {
    return R;
  }
  return [y, ...remove(x, R)];
}
```

The algorithm <em style="color:blue;">selection sort</em> is specified by two equations for an array `L`:

- If `L` is empty, the result is the empty array:
    $$ \mathtt{sort}([]) = [] $$
- Otherwise, we compute the smallest element of `L`, remove its **first** occurrence,
  sort the remainder recursively, and place the smallest element in front:
    $$ L \not= [] \rightarrow \mathtt{sort}\bigl(L\bigr) = 
       \bigl[\texttt{min}(L)\bigr] + \mathtt{sort}\bigl(\mathtt{remove}(\texttt{min}(L), L)\bigr)
    $$

```typescript
function sort(L: number[]): number[] {
  if (L.length === 0) {
    return [];
  }
  const m = Math.min(...L);
  return [m, ...sort(remove(m, L))];
}
```

```typescript
const L = [3, 5, 7, 4, 8, 1, 2, 3, 11, 13, 2];
sort(L);
```
