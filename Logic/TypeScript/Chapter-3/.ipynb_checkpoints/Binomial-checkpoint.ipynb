# Recursive Computation of the Binomial Coefficients

The *binomial coefficient* ${n \choose k}$ is defined as follows:
$$ {n \choose k} = \frac{n!}{k! \cdot (n-k)!} $$
The function `binomial(n, k)` returns ${n \choose k}$.

```typescript
def binomial(n, k):
    if k == 0 or k == n:
        return 1
    else:
        return binomial(n - 1, k - 1) + binomial(n - 1, k)                
```

```typescript
function binomial(n: number, k: number): number {
    if (!Number.isInteger(n) || !Number.isInteger(k)) 
        throw new Error("Parameter n und k müssen ganze Zahlen sein.");  
    else if (k < 0 || k > n) 
        return 0;
    else if(k === 0 || k === n)
        return 1;
    else
        return binomial(n - 1, k - 1) + binomial(n - 1, k);
}
```

```typescript
for n in range(12):
    for k in range(n+1):
        assert binomial(n, k) == math.comb(n, k)
```
